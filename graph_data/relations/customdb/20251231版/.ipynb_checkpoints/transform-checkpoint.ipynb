```python
import json
import csv
import os

# 当前目录和输出目录
# 当前目录和输出目录
current_dir = os.getcwd()
output_dir = os.path.abspath(os.path.join(current_dir, '.'))
output_path = os.path.join(output_dir, 'merged_knowledges.tsv')

# 你要保留的字段
fieldnames = [
    'relation_type',
    'source_name',
    'source_type',
    'source_id',
    'target_name',
    'target_type',
    'target_id',
    'key_sentence',
    'pmid'
]

# 要读取的 JSON 文件名
input_files = [
    'knowledges (1).json',
    'knowledges (2).json',
    'knowledges (3).json',
    'knowledges (4).json',
]

# 合并数据
data = []
for fname in input_files:
    file_path = os.path.join(current_dir, fname)
    with open(file_path, 'r', encoding='utf-8') as f:
        items = json.load(f)
        for item in items:
            row = {k: item.get(k, '') for k in fieldnames}
            data.append(row)

# 写入为 TSV 文件
with open(output_path, 'w', encoding='utf-8', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames, delimiter='\t')
    writer.writeheader()
    writer.writerows(data)

print(f"✅ 合并完成，共写入 {len(data)} 条记录")
print(f"📄 文件已保存到: {output_path}")
```
