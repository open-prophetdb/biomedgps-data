```python
import json
import csv
import os
import builtins

# =========================
# 文件设置
# =========================
current_dir = os.getcwd()

input_file = "part-00000-1c21f13f-8f8a-4584-8155-630424580672-c000.json"
output_file = "IMPC_gene_phenotype_association.tsv"

input_path = os.path.join(current_dir, input_file)
output_path = os.path.join(current_dir, output_file)

# =========================
# 输出字段
# =========================
fieldnames = [
    # 标准知识图谱关系字段
    "relation_type",
    "source_name",
    "source_type",
    "source_id",
    "source_internal_id",
    "target_name",
    "target_type",
    "target_id",
    "key_sentence",
    "pmid",
    "resource",

    # IMPC 证据字段
    "allele_accession_id",
    "mouse_allele_id",
    "p_value",
    "effect_size",
    "sex",
    "zygosity",
    "life_stage",
    "project_name",
    "phenotyping_center_id",
    "statistical_result_id",
    "parameter_id"
]

success_count = 0
error_count = 0

# =========================
# 逐行读取并写入 TSV
# =========================
with builtins.open(input_path, "r", encoding="utf-8") as input_f, \
     builtins.open(output_path, "w", encoding="utf-8", newline="") as output_f:

    writer = csv.DictWriter(
        output_f,
        fieldnames=fieldnames,
        delimiter="\t"
    )

    writer.writeheader()

    for line_number, line in enumerate(input_f, start=1):
        line = line.strip()

        if not line:
            continue

        try:
            item = json.loads(line)
        except json.JSONDecodeError as e:
            error_count += 1
            print(f"⚠️ 第 {line_number} 行解析失败：{e}")
            continue

        # phenotype 是嵌套字典
        phenotype = item.get("phenotype", {}) or {}

        phenotype_id = phenotype.get("id", "")
        phenotype_name = phenotype.get("name", "")

        # 标准 MGI 基因编号
        source_id = item.get("genePhenotypeAssociationId", "")

        # IMPC 内部基因标识符
        source_internal_id = item.get("mouseGeneId", "")

        sex = item.get("sex", "")
        zygosity = item.get("zygosity", "")
        life_stage = item.get("lifeStageName", "")

        # 自动构建可读描述
        key_sentence = (
            f"Mouse gene {source_id} is associated with "
            f"{phenotype_name} ({phenotype_id})"
        )

        if zygosity:
            key_sentence += f" in {zygosity} mice"

        if sex:
            key_sentence += f", sex: {sex}"

        if life_stage:
            key_sentence += f", life stage: {life_stage}"

        key_sentence += "."

        row = {
            # 标准关系字段
            "relation_type": "IMPC::associatied_with::Gene:Phenotype",
            "source_name": "",  # 当前文件没有基因 symbol
            "source_type": "mouse_gene",
            "source_id": source_id,
            "source_internal_id": source_internal_id,
            "target_name": phenotype_name,
            "target_type": "phenotype",
            "target_id": phenotype_id,
            "key_sentence": key_sentence,
            "pmid": "",
            "resource": "IMPC",

            # IMPC 证据字段
            "allele_accession_id": item.get("alleleAccessionId", ""),
            "mouse_allele_id": item.get("mouseAlleleId", ""),
            "p_value": item.get("pValue", ""),
            "effect_size": item.get("effectSize", ""),
            "sex": sex,
            "zygosity": zygosity,
            "life_stage": life_stage,
            "project_name": item.get("projectName", ""),
            "phenotyping_center_id": item.get("phenotypingCenterId", ""),
            "statistical_result_id": item.get("statisticalResultId", ""),
            "parameter_id": item.get("parameterId", "")
        }

        writer.writerow(row)
        success_count += 1

print("\n✅ 提取完成")
print(f"📊 成功写入：{success_count} 条记录")
print(f"⚠️ 解析失败：{error_count} 条记录")
print(f"📄 输出文件：{output_path}")
```

#### Extract gene mapping table

```python
import json
import csv
import os
import builtins

# =========================
# 文件设置
# =========================
current_dir = os.getcwd()

input_file = "gene_mouse.json"
output_file = "IMPC_mouse_gene_mapping.tsv"

input_path = os.path.join(current_dir, input_file)
output_path = os.path.join(current_dir, output_file)

# =========================
# 输出字段
# =========================
fieldnames = [
    "gene_mgi_id",
    "gene_symbol",
    "gene_full_name"
]

success_count = 0
missing_id_count = 0
duplicate_count = 0
error_count = 0

# 用于检查重复 MGI ID
seen_ids = set()

# =========================
# 逐行读取 JSONL 并写入 TSV
# =========================
with builtins.open(input_path, "r", encoding="utf-8") as input_f, \
     builtins.open(output_path, "w", encoding="utf-8", newline="") as output_f:

    writer = csv.DictWriter(
        output_f,
        fieldnames=fieldnames,
        delimiter="\t"
    )

    writer.writeheader()

    for line_number, line in enumerate(input_f, start=1):
        line = line.strip()

        if not line:
            continue

        try:
            item = json.loads(line)
        except json.JSONDecodeError as e:
            error_count += 1
            print(f"⚠️ 第 {line_number} 行解析失败：{e}")
            continue

        gene_mgi_id = item.get("mgiGeneAccessionId", "")
        gene_symbol = item.get("symbol", "")
        gene_full_name = item.get("name", "")

        # 没有标准 MGI ID 的记录无法用于映射
        if not gene_mgi_id:
            missing_id_count += 1
            continue

        # 避免重复写入同一个 MGI ID
        if gene_mgi_id in seen_ids:
            duplicate_count += 1
            continue

        seen_ids.add(gene_mgi_id)

        row = {
            "gene_mgi_id": gene_mgi_id,
            "gene_symbol": gene_symbol,
            "gene_full_name": gene_full_name
        }

        writer.writerow(row)
        success_count += 1

# =========================
# 输出统计信息
# =========================
print("\n========== 映射表生成完成 ==========")
print(f"✅ 成功提取基因：{success_count} 条")
print(f"⚠️ 缺少 MGI ID：{missing_id_count} 条")
print(f"⚠️ 重复 MGI ID：{duplicate_count} 条")
print(f"⚠️ JSON 解析失败：{error_count} 条")
print(f"📄 映射表已保存到：{output_path}")
```

##### Check the mapping for null values

```python
import pandas as pd


# =========================
# 需要检查的关键字段
# =========================
check_columns = [
    "source_id",
    "source_name",
    "target_id",
    "target_name",
    "relation_type"
]

# =========================
# 检查列是否存在
# =========================
missing_columns = [
    column for column in check_columns
    if column not in formatted_IMPC.columns
]

if missing_columns:
    raise ValueError(f"formatted_IMPC 中缺少字段：{missing_columns}")

# =========================
# 构建空值判断条件
# =========================
# 同时识别：
# 1. NaN
# 2. 空字符串 ""
# 3. 仅包含空格的字符串
# 4. 字符串形式的 "nan" 和 "None"
invalid_mask = pd.Series(False, index=formatted_IMPC.index)

for column in check_columns:
    column_as_str = formatted_IMPC[column].astype(str).str.strip()

    current_invalid = (
        formatted_IMPC[column].isna() |
        (column_as_str == "") |
        (column_as_str.str.lower() == "nan") |
        (column_as_str.str.lower() == "none")
    )

    invalid_mask = invalid_mask | current_invalid

# =========================
# 拆分数据
# =========================
unformatted_IMPC = formatted_IMPC.loc[invalid_mask].copy()
formatted_IMPC_clean = formatted_IMPC.loc[~invalid_mask].copy()

# 重置行号
unformatted_IMPC = unformatted_IMPC.reset_index(drop=True)
formatted_IMPC_clean = formatted_IMPC_clean.reset_index(drop=True)

# =========================
# 保存文件
# =========================
unformatted_IMPC.to_csv(
    "unformatted_IMPC.tsv",
    sep="\t",
    index=False,
    encoding="utf-8"
)

formatted_IMPC_clean.to_csv(
    "formatted_IMPC.tsv",
    sep="\t",
    index=False,
    encoding="utf-8"
)

# =========================
# 输出统计结果
# =========================
print("========== 拆分完成 ==========")
print(f"原始记录数：{len(formatted_IMPC)}")
print(f"✅ 完整记录数：{len(formatted_IMPC_clean)}")
print(f"⚠️ 存在空值的记录数：{len(unformatted_IMPC)}")
print("\n已保存文件：")
print("✅ formatted_IMPC.tsv")
print("⚠️ unformatted_IMPC.tsv")
```

##### Extract phenotype mappings

```python
import pandas as pd

# ============================================================
# 1. 读取两张 MP–HPO 映射表
# ============================================================
mgi_raw = pd.read_csv(
    "mp_hp_mgi_all.sssom.tsv",
    sep="\t",
    comment="#",
    dtype=str,
    keep_default_na=False
)

upheno_raw = pd.read_csv(
    "upheno-cross-species.sssom.tsv",
    sep="\t",
    comment="#",
    dtype=str,
    keep_default_na=False
)

print("MGI 人工映射原始记录数：", len(mgi_raw))
print("uPheno 跨物种映射原始记录数：", len(upheno_raw))


# ============================================================
# 2. 定义函数：统一提取 MP → HPO 映射
# ============================================================
def extract_mp_hpo(df, mapping_source):
    """
    从 SSSOM 文件中提取 MP–HPO 映射。
    无论原始方向是 MP → HP，还是 HP → MP，
    最终统一整理为 MP → HP。
    """

    work = df.copy()

    # 某些文件没有这些字段，自动补空列
    for column in [
        "subject_label",
        "object_label",
        "confidence"
    ]:
        if column not in work.columns:
            work[column] = ""

    # 清理字符串
    for column in [
        "subject_id",
        "subject_label",
        "predicate_id",
        "object_id",
        "object_label",
        "confidence"
    ]:
        work[column] = (
            work[column]
            .fillna("")
            .astype(str)
            .str.strip()
        )

    # ----------------------------------------
    # 情况 1：原始方向为 MP → HP
    # ----------------------------------------
    forward = work[
        work["subject_id"].str.match(r"^MP:\d+$", na=False)
        &
        work["object_id"].str.match(r"^HP:\d+$", na=False)
    ].copy()

    forward = forward.rename(
        columns={
            "subject_id": "mp_id",
            "subject_label": "mp_name",
            "object_id": "hpo_id",
            "object_label": "hpo_name",
            "predicate_id": "mapping_type"
        }
    )

    # ----------------------------------------
    # 情况 2：原始方向为 HP → MP
    # ----------------------------------------
    reverse = work[
        work["subject_id"].str.match(r"^HP:\d+$", na=False)
        &
        work["object_id"].str.match(r"^MP:\d+$", na=False)
    ].copy()

    reverse = reverse.rename(
        columns={
            "object_id": "mp_id",
            "object_label": "mp_name",
            "subject_id": "hpo_id",
            "subject_label": "hpo_name",
            "predicate_id": "mapping_type"
        }
    )

    # 反转方向时，broadMatch 和 narrowMatch 的含义需要互换
    reverse_mapping_type = {
        "skos:broadMatch": "skos:narrowMatch",
        "skos:narrowMatch": "skos:broadMatch"
    }

    reverse["mapping_type"] = (
        reverse["mapping_type"]
        .replace(reverse_mapping_type)
    )

    # ----------------------------------------
    # 合并两个方向
    # ----------------------------------------
    result = pd.concat(
        [
            forward[
                [
                    "mp_id",
                    "mp_name",
                    "hpo_id",
                    "hpo_name",
                    "mapping_type",
                    "confidence"
                ]
            ],
            reverse[
                [
                    "mp_id",
                    "mp_name",
                    "hpo_id",
                    "hpo_name",
                    "mapping_type",
                    "confidence"
                ]
            ]
        ],
        ignore_index=True
    )

    result["mapping_source"] = mapping_source

    result = result.drop_duplicates(
        subset=[
            "mp_id",
            "hpo_id",
            "mapping_type"
        ],
        keep="first"
    )

    return result


# ============================================================
# 3. 分别提取两张表中的 MP–HPO 映射
# ============================================================
mgi_mapping = extract_mp_hpo(
    mgi_raw,
    mapping_source="MGI_manual"
)

upheno_mapping = extract_mp_hpo(
    upheno_raw,
    mapping_source="uPheno_cross_species"
)

print("\nMGI 人工 MP–HPO 映射数：", len(mgi_mapping))
print("uPheno MP–HPO 映射数：", len(upheno_mapping))


# ============================================================
# 4. 合并两张映射表
# MGI 人工整理优先，uPheno 用于补充
# ============================================================
mp_hpo_mapping = pd.concat(
    [
        mgi_mapping,
        upheno_mapping
    ],
    ignore_index=True
)

priority = {
    "MGI_manual": 1,
    "uPheno_cross_species": 2
}

mp_hpo_mapping["source_priority"] = (
    mp_hpo_mapping["mapping_source"]
    .map(priority)
)

mp_hpo_mapping = mp_hpo_mapping.sort_values(
    [
        "mp_id",
        "hpo_id",
        "source_priority"
    ]
)

# 同一个 MP–HPO 对优先保留人工整理结果
mp_hpo_mapping = mp_hpo_mapping.drop_duplicates(
    subset=[
        "mp_id",
        "hpo_id"
    ],
    keep="first"
)

print("\n========== 合并后的映射表 ==========")
print("MP–HPO 映射总数：", len(mp_hpo_mapping))
print("可映射的不同 MP 数量：", mp_hpo_mapping["mp_id"].nunique())
print("涉及的不同 HPO 数量：", mp_hpo_mapping["hpo_id"].nunique())

print("\n映射来源统计：")
print(mp_hpo_mapping["mapping_source"].value_counts())

print("\n映射类型统计：")
print(mp_hpo_mapping["mapping_type"].value_counts(dropna=False))


# ============================================================
# 5. 保存完整 MP–HPO 映射表
# ============================================================
mp_hpo_mapping.to_csv(
    "MP_HPO_mapping_combined.tsv",
    sep="\t",
    index=False,
    encoding="utf-8"
)

print("\n✅ 已保存完整映射表：MP_HPO_mapping_combined.tsv")


# ============================================================
# 6. 准备 IMPC 关系表
# ============================================================
IMPC_work = formatted_IMPC.copy()

IMPC_work["target_id"] = (
    IMPC_work["target_id"]
    .fillna("")
    .astype(str)
    .str.strip()
)

IMPC_work["target_name"] = (
    IMPC_work["target_name"]
    .fillna("")
    .astype(str)
    .str.strip()
)

IMPC_work["key_sentence"] = (
    IMPC_work["key_sentence"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# 保存原始 MP 信息
IMPC_work["original_mp_id"] = IMPC_work["target_id"]
IMPC_work["original_mp_name"] = IMPC_work["target_name"]


# ============================================================
# 7. 根据 MP ID 映射到 HPO
# ============================================================
IMPC_mapped = IMPC_work.merge(
    mp_hpo_mapping,
    how="left",
    left_on="original_mp_id",
    right_on="mp_id"
)

IMPC_mapped["hpo_id"] = (
    IMPC_mapped["hpo_id"]
    .fillna("")
    .astype(str)
    .str.strip()
)

IMPC_mapped["hpo_name"] = (
    IMPC_mapped["hpo_name"]
    .fillna("")
    .astype(str)
    .str.strip()
)


# ============================================================
# 8. 拆分成功映射和未成功映射的记录
# ============================================================
formatted_IMPC_HPO = IMPC_mapped[
    IMPC_mapped["hpo_id"] != ""
].copy()

unmapped_IMPC_MP = IMPC_mapped[
    IMPC_mapped["hpo_id"] == ""
].copy()


# ============================================================
# 9. 将 HPO 写入 target_id 和 target_name
# ============================================================
formatted_IMPC_HPO["target_id"] = (
    formatted_IMPC_HPO["hpo_id"]
)

formatted_IMPC_HPO["target_name"] = (
    formatted_IMPC_HPO["hpo_name"]
)

# 应与 entities.tsv 中 HPO 的 label 保持一致
formatted_IMPC_HPO["target_type"] = "Phenotype"


# ============================================================
# 10. 在 key_sentence 中保留原始 MP 信息
# ============================================================
formatted_IMPC_HPO["key_sentence"] = (
    formatted_IMPC_HPO["key_sentence"]
    +
    " Original IMPC MP phenotype: "
    +
    formatted_IMPC_HPO["original_mp_name"]
    +
    " ("
    +
    formatted_IMPC_HPO["original_mp_id"]
    +
    "). Mapped HPO phenotype: "
    +
    formatted_IMPC_HPO["target_name"]
    +
    " ("
    +
    formatted_IMPC_HPO["target_id"]
    +
    "). Mapping type: "
    +
    formatted_IMPC_HPO["mapping_type"]
    +
    ". Mapping source: "
    +
    formatted_IMPC_HPO["mapping_source"]
    +
    "."
)


# ============================================================
# 11. 去重
# ============================================================
formatted_IMPC_HPO = formatted_IMPC_HPO.drop_duplicates(
    subset=[
        "source_id",
        "target_id",
        "relation_type",
        "original_mp_id",
        "mapping_type"
    ],
    keep="first"
)


# ============================================================
# 12. 输出统计结果
# ============================================================
print("\n========== IMPC 的 MP → HPO 映射结果 ==========")
print("原始 IMPC 关系数：", len(formatted_IMPC))
print("映射后生成的 HPO 关系数：", len(formatted_IMPC_HPO))
print("仍未映射的 IMPC 关系数：", len(unmapped_IMPC_MP))

print("\n原始 IMPC 中不同 MP 表型数：")
print(formatted_IMPC["target_id"].nunique())

print("\n成功映射的不同 MP 表型数：")
print(formatted_IMPC_HPO["original_mp_id"].nunique())

print("\n仍未映射的不同 MP 表型数：")
print(unmapped_IMPC_MP["original_mp_id"].nunique())


# ============================================================
# 13. 保存结果
# ============================================================
formatted_IMPC_HPO.to_csv(
    "formatted_IMPC_HPO.tsv",
    sep="\t",
    index=False,
    encoding="utf-8"
)

unmapped_IMPC_MP.to_csv(
    "unmapped_IMPC_MP.tsv",
    sep="\t",
    index=False,
    encoding="utf-8"
)

print("\n✅ 已保存成功映射结果：formatted_IMPC_HPO.tsv")
print("✅ 已保存未映射结果：unmapped_IMPC_MP.tsv")
```

```python
import pandas as pd

# ============================================================
# 1. 读取小鼠基因 → 人类基因映射表
# ============================================================
mapping_file = (
    "/Users/zhuzhixing/KG/biomedgps-data/"
    "graph_data/mapping/human_mouse_gene_mappings.tsv"
)

human_mouse_mapping = pd.read_csv(
    mapping_file,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

# ============================================================
# 2. 提取需要的字段
# ============================================================
gene_mapping = human_mouse_mapping[
    [
        "mgi_id_mouse",
        "symbol_mouse",
        "entrez_id_human",
        "symbol_human"
    ]
].copy()

# 清理空格
for column in [
    "mgi_id_mouse",
    "symbol_mouse",
    "entrez_id_human",
    "symbol_human"
]:
    gene_mapping[column] = (
        gene_mapping[column]
        .fillna("")
        .astype(str)
        .str.strip()
    )

# ============================================================
# 3. 只保留有效映射
# 注意：entrez_id_human 已经包含 ENTREZ: 前缀
# ============================================================
gene_mapping = gene_mapping[
    gene_mapping["mgi_id_mouse"]
    .str.match(r"^MGI:\d+$", na=False)
    &
    gene_mapping["entrez_id_human"]
    .str.match(r"^ENTREZ:\d+$", na=False)
].copy()

# 去重
gene_mapping = gene_mapping.drop_duplicates(
    subset=[
        "mgi_id_mouse",
        "entrez_id_human"
    ],
    keep="first"
)

print("有效 MGI → 人类 Entrez 映射数：", len(gene_mapping))
print(
    "可映射的小鼠 MGI 基因数：",
    gene_mapping["mgi_id_mouse"].nunique()
)

display(gene_mapping.head(20))

# ============================================================
# 4. 准备已完成 MP → HPO 映射的 IMPC 文件
# ============================================================
IMPC_work = formatted_IMPC_HPO.copy()

IMPC_work["source_id"] = (
    IMPC_work["source_id"]
    .fillna("")
    .astype(str)
    .str.strip()
)

IMPC_work["source_name"] = (
    IMPC_work["source_name"]
    .fillna("")
    .astype(str)
    .str.strip()
)

IMPC_work["key_sentence"] = (
    IMPC_work["key_sentence"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# 保存原始小鼠基因信息
IMPC_work["original_mgi_id"] = IMPC_work["source_id"]
IMPC_work["original_mouse_gene_name"] = IMPC_work["source_name"]

# ============================================================
# 5. 按照 MGI ID 映射到人类 Entrez 基因
# ============================================================
IMPC_gene_mapped = IMPC_work.merge(
    gene_mapping[
        [
            "mgi_id_mouse",
            "symbol_mouse",
            "entrez_id_human",
            "symbol_human"
        ]
    ],
    how="left",
    left_on="original_mgi_id",
    right_on="mgi_id_mouse"
)

# 清理映射结果
IMPC_gene_mapped["entrez_id_human"] = (
    IMPC_gene_mapped["entrez_id_human"]
    .fillna("")
    .astype(str)
    .str.strip()
)

IMPC_gene_mapped["symbol_human"] = (
    IMPC_gene_mapped["symbol_human"]
    .fillna("")
    .astype(str)
    .str.strip()
)

# ============================================================
# 6. 拆分成功和失败记录
# ============================================================
formatted_IMPC_final = IMPC_gene_mapped[
    IMPC_gene_mapped["entrez_id_human"] != ""
].copy()

unmapped_IMPC_gene = IMPC_gene_mapped[
    IMPC_gene_mapped["entrez_id_human"] == ""
].copy()

# ============================================================
# 7. 同时替换 source_id 和 source_name
# ============================================================
formatted_IMPC_final["source_id"] = (
    formatted_IMPC_final["entrez_id_human"]
)

formatted_IMPC_final["source_name"] = (
    formatted_IMPC_final["symbol_human"]
)

formatted_IMPC_final["source_type"] = "Gene"

# ============================================================
# 8. 将原始小鼠基因信息追加到 key_sentence
# ============================================================
formatted_IMPC_final["key_sentence"] = (
    formatted_IMPC_final["key_sentence"]
    +
    " Original IMPC mouse gene: "
    +
    formatted_IMPC_final["original_mouse_gene_name"]
    +
    " ("
    +
    formatted_IMPC_final["original_mgi_id"]
    +
    "). Mapped human ortholog: "
    +
    formatted_IMPC_final["source_name"]
    +
    " ("
    +
    formatted_IMPC_final["source_id"]
    +
    ")."
)

# ============================================================
# 9. 删除辅助列
# ============================================================
formatted_IMPC_final = formatted_IMPC_final.drop(
    columns=[
        "mgi_id_mouse",
        "symbol_mouse",
        "entrez_id_human",
        "symbol_human"
    ],
    errors="ignore"
)

# ============================================================
# 10. 去重
# ============================================================
formatted_IMPC_final = formatted_IMPC_final.drop_duplicates(
    subset=[
        "source_id",
        "target_id",
        "relation_type",
        "original_mgi_id",
        "original_mp_id"
    ],
    keep="first"
)

# ============================================================
# 11. 保存结果
# ============================================================
formatted_IMPC_final.to_csv(
    "formatted_IMPC_final.tsv",
    sep="\t",
    index=False,
    encoding="utf-8"
)

unmapped_IMPC_gene.to_csv(
    "unmapped_IMPC_gene.tsv",
    sep="\t",
    index=False,
    encoding="utf-8"
)

print("\n========== 映射结果 ==========")
print("输入 IMPC–HPO 关系数：", len(formatted_IMPC_HPO))
print("映射成功的关系数：", len(formatted_IMPC_final))
print("未成功映射的关系数：", len(unmapped_IMPC_gene))

print(
    "成功映射的小鼠 MGI 基因数：",
    formatted_IMPC_final["original_mgi_id"].nunique()
)

print(
    "映射后的人类 Entrez 基因数：",
    formatted_IMPC_final["source_id"].nunique()
)

print("\n✅ 已保存：formatted_IMPC_final.tsv")
print("✅ 已保存：unmapped_IMPC_gene.tsv")
```

#### formatted IMPC

```python
import os
import os.path as osp
import subprocess


def format_IMPC(filename):
    def get_project_root():
        try:
            return osp.dirname(osp.dirname(os.getcwd()))
        except Exception as e:
            raise RuntimeError(f"Failed to determine project root: {e}")

    try:
        root_dir = get_project_root()
        print(f"Project root directory: {root_dir}")
    except RuntimeError as e:
        print(e)
        exit(1)

    database = "customdb"
    relations_path = osp.join(
        root_dir,
        "relations",
        "IMPC",
        filename,
    )
    output_dir = osp.join(
        root_dir, "formatted_relations", "IMPC"
    )
    entities_path = osp.join(root_dir, "entities.tsv")
    log_file = osp.join(output_dir, "log.txt")
    relation_types_file = osp.join(root_dir, "relation_types.tsv")

    command = [
        "graph-builder",
        "--database",
        database,
        "-d",
        relations_path,
        "-o",
        output_dir,
        "-f",
        entities_path,
        "-n",
        "20",
        "--download",
        "--skip",
        "-l",
        log_file,
        "--debug",
        "--relation-type-dict-fpath",
        relation_types_file,
    ]

    print("Executing command:", " ".join(command))

    try:
        subprocess.run(command, check=True)
    except FileNotFoundError:
        print(
            "Error: 'graph-builder' command not found. Make sure it is installed and available in the PATH."
        )
        exit(1)
    except subprocess.CalledProcessError as e:
        print(f"Error: Command execution failed with return code {e.returncode}")
        print(f"Output: {e.output}")
        exit(1)
    except Exception as e:
        print(f"Unexpected error: {e}")
        exit(1)
```
